```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "http://books.toscrape.com/catalogue/category/books_1/index.html"
headers = {'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_15_7)'}

response = requests.get(url, headers=headers)
```

```python
# PARTE 1: WEB SCRAPING
if response.status_code == 200:
    soup = BeautifulSoup(response.content, 'html.parser')
    livros = []

    for artigo in soup.find_all('article', class_='product_pod'):
        titulo = artigo.h3.a['title']
        preco_bruto = artigo.find('p', class_='price_color').text

        livros.append({
            'titulo': titulo,
            'preco_bruto': preco_bruto,
            'data_coleta': '2026-10-07'
        })

    df = pd.DataFrame(livros)
    print("1. Dados coletados com sucesso!")
```

```python
# PARTE 2 e 3: LIMPEZA E TRATAMENTO (Substitui o OpenRefine)
# 1. Limpeza do preço (remove '£', remove espaços extras e converte para float)
df['preco'] = df['preco_bruto'].str.replace('£', '', regex=False).str.strip().astype(float)

# 2. Remoção de registros duplicados com base no título
df.drop_duplicates(subset=['titulo'], inplace=True)

# 3. Tratamento de valores nulos (se houver)
df['preco'] = df['preco'].fillna(0)

# 4. Padronização do formato de data
df['data_coleta'] = pd.to_datetime(df['data_coleta'], format='%Y-%m-%d')

# 5. Filtragem de dados (exemplo: produtos com preço acima de 20.00)
df_limpo = df[df['preco'] > 20.00].copy()

# 6. Salvando o resultado final
df_limpo.to_csv('dados_limpos.csv', index=False, encoding='utf-8')
print("2. Limpeza concluída e arquivo 'dados_limpos.csv' salvo sem usar OpenRefine!")
```
